```r
# Define the file path
file_path <- "~/Downloads/half-marathon.txt"

# Read the file
data <- readLines(file_path,encoding = "UTF-8")
# Change the encoding to fix accent issues
data <- iconv(data, from = "UTF-8", to = "latin1")
# Print the content of the file
print(data)
```

```r
library(rvest)

# Parse the HTML content
html_content <- paste(data, collapse = "\n")
parsed_html <- read_html(html_content)

# Extract the relevant information
rows <- parsed_html %>%
    html_nodes(".testimonial") %>%
    lapply(function(node) {
        name <- node %>% html_node("h6") %>% html_text(trim = TRUE)
        details <- node %>% html_nodes("li") %>% html_text(trim = TRUE)
        c(Name = name, details)
    })

# Convert to a data frame
df <- do.call(rbind, lapply(rows, function(row) {
    data.frame(matrix(row, ncol = length(row), byrow = TRUE), stringsAsFactors = FALSE)
}))
colnames(df) <- c("Name", "Genero", "BIB", "Tiempo_Chip", "Paso_Medio", "Categoria", "Tiempo_Oficial")

# Print the data frame
print(df)
```
